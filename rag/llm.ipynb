**1) Import libraries**

```python
from google import genai

print("Gemini SDK imported successfully!")
```

google is the package namespace.



genai is the module that gives us access to Gemini.



We import it so we can configure a Gemini client and send requests to a model.

**2) Load the API key**

```python
import os
from pathlib import Path
from dotenv import load_dotenv
from google import genai

# Locate the .env file in the parent folder
env_path = Path.cwd().parent / ".env"

# Load environment variables
load_dotenv(dotenv_path=env_path)

# Get the API key
api_key = os.getenv("GEMINI_API_KEY")

# Check if the key was loaded
if api_key:
    print("API key loaded successfully!")
else:
    print("API key not found. Check your .env file.")
```

**import os**	Access environment variables

**load_dotenv()**	Read variables from .env

**os.getenv()**	Retrieve the API key

**if api_key**	Check whether the key exists


**3) Configure the Gemini client**

```python
# Create the Gemini client
client = genai.Client(api_key=api_key)

print("Gemini client created successfully!")
```

API key: authenticates your access.



Gemini client: represents the connection between your Python program and Google's API.



Gemini model: the AI that will generate responses.

**genai.Client()** creates the connection object.



**api_key=api_key** gives it the credential we loaded from .env.



**client** is the object we'll use later to send questions to Gemini.

**4) Test the connection**

```python
MODEL_NAME = "gemini-3.5-flash-lite"

response = client.models.generate_content(
    model=MODEL_NAME,
    contents="Say hello and confirm that the connection works."
)

print(response.text)
```

- MODEL_NAME selects the Gemini model assigned to your project.
- generate_content() sends a request to Gemini.
- response.text displays the generated answer.


This uses the official SDK's generate_content() method
