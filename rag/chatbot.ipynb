# Pharmaly Chatbot User Interface

This notebook provides a simple interactive interface for users to communicate with the Pharmaly chatbot.

Users can ask questions about Pharmaly products, receive answers with sources, and end the conversation by typing `exit`.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

# Remove the notebook's current directory from Python's import paths
sys.path = [
    p for p in sys.path
    if Path(p or ".").resolve() != Path.cwd().resolve()
]

# Put the project root first
sys.path.insert(0, str(PROJECT_ROOT))

from rag.rag import ask
from rag import config

history = []

print("RAG imported successfully!")
print("Project root:", PROJECT_ROOT)
print("Threshold:", config.SIMILARITY_THRESHOLD)
```

```python
print("Welcome to Pharmaly Chatbot!")
print("Ask me questions about Pharmaly products.")
print("Type 'exit' to end the conversation.")

while True:

    question = input("\nYou: ").strip()

    if question.lower() in {"exit", "quit"}:
        print("Pharmaly: Goodbye!")
        break

    if not question:
        print("Pharmaly: Please enter a question.")
        continue
        print("\nQuestion:", question)
    try:
        result = ask(question, history)
        print('\nQuestion:', question)
        print("\nPharmaly:", result["answer"])

        if result["sources"]:
            print("\nSources:")
            for source in result["sources"]:
                print("-", source["name"])
                print(" ", source["url"])

        history.append((question, result["answer"]))
        history = history[-config.MAX_HISTORY_TURNS:]

    except Exception as exc:
        print(f"\n[error] {exc}")
```
