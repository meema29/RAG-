# Pharmaly Chatbot — User Interface

This notebook provides a simple interactive interface for users to communicate with the Pharmaly chatbot.

Users can ask questions about Pharmaly products, receive answers with sources, and end the conversation by typing `exit`.

```python
import importlib
import rag_pipeline

importlib.reload(rag_pipeline)

rag_answer = rag_pipeline.rag_answer
```

```python
print("Welcome to Pharmaly Chatbot!")
print("Ask me questions about Pharmaly products.")
print("Type 'exit' to end the conversation.")

while True:

    question = input("\nYou: ")

    if question.lower().strip() == "exit":
        print("Pharmaly: Goodbye!")
        break

    if not question.strip():
        print("Pharmaly: Please enter a question.")
        continue

    result = rag_answer(question)

    print("\nPharmaly:", result["answer"])

    if result["sources"]:
        print("\nSources:")

        for source in result["sources"]:
            print("-", source["product_name"])
            print(" ", source["url"])
```
